<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Comenzamos con las <b>normas vectoriales</b>.
    <br>
    Recordamos que en $\mathbb{K}^n$, para cada $p\ge1$, las funciones
\[
\|\cdot\|_p \colon \mathbb{K}^n \to \mathbb{R}^+ \,/\, X \mapsto \|X\|_p = \left( \sum_{i=1}^n |x_i|^p \right)^{1/p}\,,
\]
    así como la función
\[
\|\cdot\|_\infty \colon \mathbb{K}^n \to \mathbb{R}^+ \,/\, X \mapsto \|X\|_\infty = \max_{i=1,2,\ldots,n} |x_i|\,,
\]
    definen normas vectoriales, llamadas <b>normas-</b>$p$. Además se sabe que para todo $X\in\mathbb{K}^n$ la sucesión $\{\|X\|_p\}_{p\in\mathbb{N}}$ es convergente y su límite es justamente $\|X\|_\infty$.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    La <b>biblioteca</b> <span style="font-family: Courier">Numpy</span> incluye la <b>función</b> <span style="font-family: Courier">norm()</span> cuyo objetivo es calcular normas vectoriales (también normas matriciales como veremos más adelante); cuando el <b>primer argumento de entrada</b> de dicha función es un vector (array de una dimensión), calcula la norma indicada por el <b>segundo argumento de entrada</b>; dicho segundo argumento puede ser (entre otros) cualquier número real $p\ge1$ o bien la cadena de caracteres <span style="font-family: Courier">inf</span>.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Trabajamos ahora con las <b>normas matriciales</b>.
    <br>
    Recordamos que es posible definir una norma matricial en $\mathcal{M}_n(\mathbb{K})$ a partir de una norma vectorial en $\mathbb{K}^n$ poniendo
\[
\|A\| = \max_{X\in\mathbb{K}^n-\{0\}} \frac{\|A\,X\|}{\|X\|} = \max_{\|X\|=1} \|A\,X\| = \max_{\|X\|\le1} \|A\,X\|\,;
\]
    esta norma matricial recibe el nombre de <b>norma matricial subordinada a la norma vectorial</b> considerada, y en el caso de la normas vectoriales más habituales se tienen las caracterizaciones
\begin{eqnarray*}
\|A\|_\infty & = & \max_{i=1,\ldots,n} \sum_{j=1}^n |a_{i,j}|\,, \\
\|A\|_1 & = & \max_{j=1,\ldots,n} \sum_{i=1}^n |a_{i,j}|\,, \\
\|A\|_2 & = & \sqrt{\varrho(A^*\,A)} = \sqrt{\varrho(A\,A^*)}\,;
\end{eqnarray*}
    recordamos también que la norma matricial-2 recibe el nombre de <b>norma espectral</b>.
    <br>
    Por otro lado, la función
\[
\|\cdot\|_E \colon \mathcal{M}_n(\mathbb{K}) \to \mathbb{R}^+ \,/\, A \mapsto \|A\|_E = \left( \sum_{i=1}^n \sum_{j=1}^n |a_{i,j}|^2 \right)^{1/2}
\]
define también una norma matricial, no subordinada, llamada <b>norma matricial euclídea, de Shur o de Frobenius</b>.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Como ya sabemos, la <b>biblioteca</b> <span style="font-family: Courier">Numpy</span> incluye la <b>función</b> <span style="font-family: Courier">norm()</span> que permite también calcular normas matriciales; cuando el <b>primer argumento de entrada</b> de dicha función es una matriz (array de dos dimensiones), calcula la norma indicada por el <b>segundo argumento de entrada</b>; dicho segundo argumento puede ser (entre otros) <span style="font-family: Courier">inf</span>, 1, 2 o <span style="font-family: Courier">'fro'</span>.
    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    <span style="color:#FF0000"><b>Ejercicio 4.</b></span> Definir una función, de nombre <span style="font-family: Courier">norma_mat()</span>, con dos argumentos de entrada: el primero una matriz cuadrada, y el segundo el índice $p$ de la norma matricial que se quiere calcular (este índice puede adoptar los valores <span style="font-family: Courier">inf</span>, 1, 2 o <span style="font-family: Courier">'fro'</span>); esta función debe devolver la norma matricial correspondiente.
    </span></div>

```python
def norma_mat(A, p):
    nd = ndim(A)
    if nd == 2:#me han pasado una matriz, sino es una matriz entonces no puede ser una matriz y fuera
        m, n = shape(A)#me da el num de fila y num de colu
    if nd != 2 or m != n:
        return "Error norma_mat: tipo de A."
    AA = array(A, dtype=complex)
    if p == inf:
        return max(sum(abs(AA),axis=1))#hace la operacion sum pero con el eje 1 q son las filas, lo hago pq sino me hace sum de todo y quiero q me lo haga por filas
    elif p == 1:
        return max(sum(abs(AA),axis=0))
    elif p == 2:
        return max(svd(AA)[1]) #me quedo con la 2º que me devuelve svd
#         return ...
    elif p == 'fro':
        return sqrt(sum(abs(AA)**2))
    else:
        return "Error norma_mat: valor de p."
```
