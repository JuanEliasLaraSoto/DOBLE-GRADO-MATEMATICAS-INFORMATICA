<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Para definir un <b>vector</b> (fila) o una <b>matriz</b> utilizamos la <b>función</b> <span style="font-family: Courier">array()</span> cuyo <b>primer argumento</b> son los elementos del <span style="font-family: Courier">array</span>, separados por comas, encerrados entre corchetes y, en el caso de las matrices, las filas encerradas entre otros corchetes y también separadas por comas. La <b>función</b> <span style="font-family: Courier">print()</span> permite imprimirlos en pantalla. La <b>clase</b> a la que pertenecen (<span style="font-family: Courier">numpy.ndarray</span>) se obtiene con la <b>función</b> <span style="font-family: Courier">type()</span>, el <b>número de dimensiones</b> (1 en el caso de vectores y 2 en el de matrices) con la <b>función</b> <span style="font-family: Courier">ndim()</span> (también se puede utilizar el <b>comando</b> <span style="font-family: Courier">.ndim</span>), el <b>tamaño</b> (longitud del vector, o filas y columnas de las matrices) con la <b>función</b> <span style="font-family: Courier">shape()</span> (también se puede utilizar el <b>comando</b> <span style="font-family: Courier">.shape</span>), mientras que el <b>número total de elementos</b> (longitud del vector, o producto del número de filas por el de columnas de las matrices) con la <b>función</b> <span style="font-family: Courier">size()</span> (también se puede utilizar el <b>comando</b> <span style="font-family: Courier">.size</span>).
     </span></div>

```python
# Construcción de matrices.
A = array([[1, -2, 3], [-4, 5, -6]])#[[1fila],[2fila]]
x = array([7., -8, 9])#vector
X = array([[7, -8e0, 9]])#vector fila visto como matriz
Y = array([[1j], [-2]])
```

```python
print("Impresión de las matrices.")
print("A = \n", A)
print("x = ", x)
print("X = \n", X)
print("Y = \n", Y)
```

```python
print("Tipo de las matrices.")
print("type(A) = ", type(A))
print("type(x) = ", type(x))
print("type(X) = ", type(X))
print("type(Y) = ", type(Y))
```

```python
print("Número de dimensiones de las matrices.")#dim 2 =>matriz(o es un vector pero metido en una matriz, como pasa  con los vect colus y filas), dim 1=>vector
print("ndim(A) = ", ndim(A))
print("ndim(x) = ", ndim(x))
print("ndim(X) = ", ndim(X))
print("ndim(Y) = ", ndim(Y))
print(A.ndim, x.ndim, X.ndim, Y.ndim)
```

```python
print("Tamaño de las matrices.")#dice matriz de tam 2x3 ,1x3 o si es dim 1 pues tam 3
print("shape(A) = ", shape(A))
print("shape(x) = ", shape(x))
print("shape(X) = ", shape(X))
print("shape(Y) = ", shape(Y))
print(A.shape, x.shape, X.shape, Y.shape)
```

```python
print("Número de elementos de las matrices.")
print("size(A) = ", size(A))
print("size(x) = ", size(x))
print("size(X) = ", size(X))
print("size(Y) = ", size(Y))
print(A.size, x.size, X.size, Y.size)#es  lo mismo hacer size(x) que x.size, es lo mismo
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Se puede <b>crear</b> un <span style="font-family: Courier">array</span> lleno de 0’s con la <b>función</b> <span style="font-family: Courier">zeros()</span> y lleno de 1’s con la <b>función</b> <span style="font-family: Courier">ones()</span>; en ambos casos se debe pasar como <b>argumento</b> el tamaño (<span style="font-family: Courier">shape</span>) del <span style="font-family: Courier">array</span>. Para generar un <span style="font-family: Courier">array</span> con <i>elementos aleatorios</i> siguiendo una distribución uniforme sobre el intervalo [0, 1), la <b>función</b> es <span style="font-family: Courier">random.rand()</span>; esta función lleva tantos <b>argumentos</b> como dimensiones se quiere que tenga el <span style="font-family: Courier">array</span>, con los tamaños correspondientes. La <b>función</b> <span style="font-family: Courier">eye()</span> permite crear una matriz, no necesariamente cuadrada, cuyos elementos son todos 0, excepto los de la diagonal principal que valen 1; si esta función lleva un único <b>argumento</b> se crea una matriz cuadrada de dicho tamaño y si lleva dos una matriz rectangular; esta funcion admite un <b>argumento opcional</b> en la forma <span style="font-family: Courier">k=</span> que permite que la diagonal de 1’s esté <span style="font-family: Courier">k</span> posiciones por encima o por debajo (dependiendo del signo) de la diagonal principal.

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Las <b>funciones</b> <span style="font-family: Courier">arange()</span> y <span style="font-family: Courier">linspace()</span> permiten <b>crear vectores</b> (<span style="font-family: Courier">array</span> de una dimensión) <i>interpolando</i> valores intermedios entre dos extremos; los tres <b>argumentos</b> de la función <span style="font-family: Courier">arange</span> son el extremo inicial del intervalo (cerrado), el extremo final (abierto) y el paso de malla; los tres <b>argumentos</b> de la función <span style="font-family: Courier">linspace</span> son el extremo inicial del intervalo (cerrado), el extremo final (cerrado) y el número total de puntos interpolados. Se pueden <b>crear matrices</b> (<span style="font-family: Courier">array</span> de dos dimensiones) combinando estas funciones con la <b>función</b> <span style="font-family: Courier">reshape()</span>, cuyo primer <b>argumento</b> es un vector y el segundo el tamaño de la matriz que se creará.
    </span></div>

```python
print("Matriz creada por interpolación.")
h = arange(0, 12, 1) #arange me da lista de num [0...12) , ojo q el 0 es incluido y el 12 no es incluido , eso me dice el 1
print("h = ", h)
H = reshape(h, (3, 4))
print("H = \n", H)
print("H.dtype = ", H.dtype)
```

```python
print("Matriz creada por interpolación.")
k = linspace(0, 11, 12)#cojo del 0 al 11 y hago 12 puntos
print("k = ", k)
K = reshape(k, (4, 3))
print("K = \n", K)
print("K.dtype = ", K.dtype)
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    Las <b>operaciones elementales</b> (<b>sumas</b> con el <b>operador</b> <span style="font-family: Courier">+</span>, <b>productos</b> con el <b>operador</b> <span style="font-family: Courier">*</span> y <b>potencias</b> con el <b>operador</b> <span style="font-family: Courier,">**</span>) actúan sobre los <span style="font-family: Courier">array</span> elemento a elemento (incluido el producto y la potencia). Indicamos que tanto la suma como el producto admiten dos <b>generalizaciones</b>; por un lado sumar o multiplicar una constante, en cuyo caso suma o multiplica la constante a todos sus elementos (en el caso del producto es el usual producto por escalares); por otro lado sumar o multiplicar un vector (o matriz fila) o una matriz columna a una matriz, en cuyo caso lo suma o multiplica, elemento a elemento, a todas las filas o columnas de la matriz, siempre que las dimensiones lo permiten.
    <br>
    Para utilizar el <b>producto matricial</b> habitual debe utilizarse el <b>operador</b> <span style="font-family: Courier">@</span> y para la <b>potencia de matrices</b> la <b>función</b> <span style="font-family: Courier">matrix_power()</span>.

    </span></div>

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">

Un comportamiento totalmente análogo, incluido el argumento opcional, tienen las <b>funciones</b> <span style="font-family: Courier">max()</span> y <span style="font-family: Courier">min()</span>.
